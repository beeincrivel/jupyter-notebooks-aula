### Questão 1: Orçamento de viagem (1,0 ponto)

Escreva um programa que calcule o custo de uma viagem de carro.

O programa deve ler do usuário, usando `input`:

- a distância da viagem, em km (`float`);
- o consumo médio do carro, em km/litro (`float`);
- o preço do litro do combustível, em reais (`float`);
- o valor total dos pedágios, em reais (`float`);
- o número de pessoas que dividirão a despesa (`int`).

Em seguida o programa deve calcular e exibir:

1. a quantidade de litros necessária (distância dividida pelo consumo);
2. o custo do combustível;
3. o custo total (combustível + pedágios);
4. o custo por pessoa;
5. o tipo (`type`) das variáveis da distância e do número de pessoas.

Todos os valores em reais devem ser exibidos com **2 casas decimais** (use `round` ou formatação de string).

Exemplo:

```
Distância da viagem (km): 450
Consumo médio (km/l): 12.5
Preço do litro (R$): 5.89
Total de pedágios (R$): 32.40
Número de pessoas: 3

Litros necessários: 36.0
Custo do combustível: R$ 212.04
Custo total da viagem: R$ 244.44
Custo por pessoa: R$ 81.48
Tipo da distância: <class 'float'> | Tipo do número de pessoas: <class 'int'>
```

````python
def calcular_exerc1():
    distancia_km = float(input("Distância da viagem (km):"))
    consumo_medio_kml = float(input("Consumo médio (km/l):"))
    preco_litro = float(input("Preço do litro (R$):"))
    total_pedagios = float(input("Total de pedágios (R$):"))
    numero_pessoas = int(input("Número de pessoas:"))

    qntd_litros_necessarios = distancia_km/consumo_medio_kml
    custo_combustivel = qntd_litros_necessarios * preco_litro
    custo_total = custo_combustivel + total_pedagios
    custo_por_pessoa = custo_total/numero_pessoas

    print(f"\nLitros necessários: {qntd_litros_necessarios:.1f}")
    print(f"Custo do combustível: R$ {custo_combustivel:.2f}")
    print(f"Custo total da viagem: R$ {custo_total:.2f}")
    print(f"Custo por pessoa: R$ {custo_por_pessoa:.2f}")
    print(f"Tipo da distância: {type(distancia_km)} | Tipo do número de pessoas: {type(numero_pessoas)}")

calcular_exerc1 ()
````

### Questão 3: Distância entre dois pontos (1,0 ponto)

Usando a instrução `import` e o módulo `math`, escreva um programa que:

1. leia as coordenadas de dois pontos do plano: `x1`, `y1`, `x2` e `y2` (todos `float`);
2. calcule `dx` e `dy` como as diferenças **em valor absoluto** entre as coordenadas;
3. calcule a distância entre os pontos com a fórmula abaixo:

$$distancia = \sqrt{dx^2 + dy^2}$$

4. calcule a área do círculo que tem essa distância como raio;
5. exiba `dx`, `dy`, a distância e a área, todos arredondados para 2 casas decimais com `round`.

Exemplo:

```
x1: 1
y1: 2
x2: 4
y2: 6

dx = 3.0
dy = 4.0
Distância entre os pontos: 5.0
Área do círculo de raio 5.0: 78.54
```

````python
# Inclua aqui a resposta
import math

def calcular_exerc3():
    x1 = float(input("x1: "))
    y1 = float(input("y1: "))
    x2 = float(input("x2: "))
    y2 = float(input("y2: "))

    dx = abs(x2 - x1)
    dy = abs(y2 - y1)

    distancia = math.sqrt(dx ** 2 + dy ** 2)
    area_circulo = math.pi * (distancia ** 2)

    print(f"\ndx = {round(dx, 2)}")
    print(f"dy = {round(dy, 2)}")
    print(f"Distância entre os pontos: {round(distancia, 2)}")
    print(f"Área do círculo de raio {round(distancia, 2)}: {round(area_circulo, 2)}")

calcular_exerc3()
````

**(a)** Escreva uma função chamada `poligono(lados, tamanho)` que desenhe um polígono regular com o número de lados e o comprimento de lado informados. A função deve ter docstring.

**(b)** Escreva uma função chamada `flor(petalas, lados, tamanho)` que **reutilize** `poligono` para desenhar
uma flor.

**(c)** Faça três desenhos: um triângulo, um hexágono e uma flor.

Exemplo:

```
make_turtle()
flor(8, 4, 30)
```

````python
# (a) e (b) defina aqui as funções poligono e flor
import jupyturtle as turtle

def poligono(lados, tamanho):
    """Desenha um polígono regular com o número de lados e o comprimento de lado informados.
    Parâmetros:
    lados (int): O número de lados do polígono.
    tamanho (int): O comprimento de cada lado do polígono.
    """
    angulo = 360 / lados
    for _ in range(lados):
      turtle.forward(tamanho)
      turtle.right(angulo)

def flor(petalas, lados, tamanho):
    """Desenha uma flor com o número de petalas, lados e comprimento de lado informados.
    Parâmetros:
    petalas (int): O número de petalas da flor.
    lados (int): O número de lados do polígono.
    tamanho (int): O comprimento de cada lado do polígono.
    """
    angulo_interior = 360 / petalas
    for _ in range(petalas):
      poligono(lados, tamanho)
      turtle.right(180 - angulo_interior)
````

````python
# (c) triângulo
turtle.make_turtle()
poligono (3, 60)
````

````python
# (c) hexágono
turtle.make_turtle()
poligono (6, 40)
````

````python
# (c) flor
turtle.make_turtle()
flor(8, 4, 30)
````

### Questão 8: Condicionais encadeadas e aninhadas (1,0 ponto)

Escreva uma função chamada `classificar_imc(peso, altura)` que:

1. **valide os dados**: se `peso` **ou** `altura` for menor ou igual a zero, a função deve
   retornar `None`, sem calcular nada;
2. caso os dados sejam válidos, calcule o IMC com a fórmula `imc = peso / altura ** 2`;
3. use uma **condicional encadeada** (`if`/`elif`/`else`) para **retornar** a classificação correspondente:

| IMC | Classificação |
|---|---|
| menor que 18,5 | `"Abaixo do peso"` |
| de 18,5 a 24,9 | `"Peso normal"` |
| de 25,0 a 29,9 | `"Sobrepeso"` |
| 30,0 ou mais | `"Obesidade"` |

Depois, escreva um programa que leia o peso e a altura do usuário, chame a função e:

- se o retorno for `None`, exiba a mensagem `"Dados inválidos: peso e altura devem ser positivos."`;
- caso contrário, exiba o IMC arredondado com 2 casas decimais e a classificação.

A função deve ter **docstring**.

Exemplo:

```
Peso (kg): 72.5
Altura (m): 1.75
IMC: 23.67 -> Peso normal
```

```
Peso (kg): 70
Altura (m): 0
Dados inválidos: peso e altura devem ser positivos.
```

````python
#inclua aqui a respota
def classificar_imc(peso, altura):
    """Calcula o IMC com base no peso e altura e retorna a classificacao correspondente.
    Parâmetros:
    peso (float): O peso em quilogramas.
    altura (float): A altura em metros.
    """
    if peso <= 0 or altura <= 0:
        return None

    imc = peso / (altura ** 2)

    if imc < 18.5:
        classificacao = "Abaixo do peso"
    elif imc < 25:
        classificacao = "Peso normal"
    elif imc < 30:
        classificacao = "Sobrepeso"
    else:
        classificacao = "Obesidade"

    return round(imc, 2), classificacao

def programa_principal():
    """Lê os dados do usuário, chama a função classificar_imc e exibe o resultado."""
    # Substitui a vírgula por ponto para permitir a conversão para float
    peso_str = input("Peso (kg): ").replace(',', '.')
    altura_str = input("Altura (m): ").replace(',', '.')

    peso = float(peso_str)
    altura = float(altura_str)

    resultado = classificar_imc(peso, altura)

    if resultado is None:
        print("Dados inválidos: peso e altura devem ser positivos.")
    else:
        print(f"IMC: {resultado[0]} -> {resultado[1]}")

programa_principal()
````

````python
# Versão 3 - versão final, com as estatísticas da turma
def media_notas(nota1, nota2, nota3):
    """Calcula a média aritmética de três notas.
    Parâmetros:
        nota1 (float): primeira nota
        nota2 (float): segunda nota
        nota3 (float): terceira nota

    Retorna:
        float: a média das três notas
    """
    return (nota1 + nota2 + nota3) / 3

def situacao(media):
    """Determina a situação do aluno com base na média.
    Parâmetros:
        media (float): média do aluno

    Retorna:
        str: "Aprovado" se media >= 7,
             "Recuperação" se 4 <= media < 7,
             "Reprovado" se media < 4
    """
    if media >= 7:
        return "Aprovado"
    elif media >= 4:
        return "Recuperação"
    else:
        return "Reprovado"

quantidade = int(input("Quantos alunos? "))

print()

soma_medias = 0
aprovados = 0
maior_media = -1
melhor_aluno = ""

alunos = []

for i in range(1, quantidade + 1):
    nome = input(f"Aluno {i} - nome: ")
    nota1 = float(input("Nota 1: "))
    nota2 = float(input("Nota 2: "))
    nota3 = float(input("Nota 3: "))

    media = media_notas(nota1, nota2, nota3)
    sit = situacao(media)

    alunos.append((nome, media, sit))

    soma_medias += media
    if sit == "Aprovado":
        aprovados += 1
    if media > maior_media:
        maior_media = media
        melhor_aluno = nome

#boletim
print()
print("===== BOLETIM =====")
for nome, media, sit in alunos:
    print(f"{nome}: {media:.1f} - {sit}")
print("===================")

#Estatísticas
media_geral = soma_medias / quantidade
percentual_aprovados = (aprovados / quantidade) * 100

print(f"Média geral da turma: {media_geral:.2f}")
print(f"Aprovados: {aprovados} de {quantidade} ({percentual_aprovados:.2f}%)")
print(f"Melhor média: {melhor_aluno} ({maior_media:.2f})")
````
